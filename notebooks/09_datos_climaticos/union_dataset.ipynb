```python
import pandas as pd
import geopandas as gpd
```

```python
#día
lds_clima_1= gpd.read_file("../../data/mediciones_clima/LDS_clima.geojson")
rdp_m_clima_1= gpd.read_file("../../data/mediciones_clima/RDP-MONTES_clima.geojson")
rn_upm1_clima_1= gpd.read_file("../../data/mediciones_clima/RN-UPM1_clima.geojson")
rn_upm2_clima_1= gpd.read_file("../../data/mediciones_clima/RN-UPM2_clima.geojson")
```

```python
#2días
lds_clima_2= gpd.read_file("../../data/mediciones_clima/LDS_clima_48.geojson")
rdp_m_clima_2= gpd.read_file("../../data/mediciones_clima/RDP-MONTES_clima_48.geojson")
rn_upm1_clima_2= gpd.read_file("../../data/mediciones_clima/RN-UPM1_clima_48.geojson")
rn_upm2_clima_2= gpd.read_file("../../data/mediciones_clima/RN-UPM2_clima_48.geojson")
```

```python
#dataset: 
data_set = pd.read_csv("../../data/dataset_final/indices_acolite.csv")
```

```python
print(lds_clima_1.shape)
print(rdp_m_clima_1.shape)
print(rn_upm1_clima_1.shape)
print(rn_upm2_clima_1.shape)
```

```python
clima_1 = pd.concat([lds_clima_1, rdp_m_clima_1, rn_upm1_clima_1, rn_upm2_clima_1], ignore_index=True)
clima_1.shape
```

```python
clima_1
```

```python
clima_1.groupby("grupo_nombre").size()
```

```python
data_set_gdf = gpd.GeoDataFrame(data_set, geometry=gpd.points_from_xy(data_set["lon"], data_set["lat"]), crs="EPSG:4326")
data_set_gdf.head()
```

```python
data_set_gdf["dia"] = pd.to_datetime(data_set_gdf["fecha"]).dt.normalize()
data_set_gdf
```

```python
clima_1_sel = clima_1[["grupo_nombre", "dia", "nombre", "temperatura_minima", "temperatura_maxima", "pluviometro"]].drop_duplicates()
clima_1_sel.shape
```

```python
data_set_clima = data_set_gdf.merge(clima_1_sel, left_on=["grupo_nombre", "dia"], right_on=["grupo_nombre", "dia"], how="left")
data_set_clima[["fecha", "dia", "grupo_nombre", "nombre", "temperatura_minima", "temperatura_maxima", "pluviometro"]].head()
```

```python
data_set_clima[["nombre", "temperatura_minima", "temperatura_maxima", "pluviometro"]].isna().sum()
```
