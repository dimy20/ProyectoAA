```python
matchups = gpd.read_file("../../data/registros_limpios_matchups_nuevos.json")
```

```python
matchups
```

```python
matchups.groupby("grupo_nombre").size()
```

```python
puntos = matchups.drop_duplicates(subset="geometry")
puntos = puntos[["grupo_nombre", "geometry"]]
```

```python
puntos
```

```python
puntos.groupby("grupo_nombre").size()
```

```python
estaciones = clima.drop_duplicates(subset="nombre")
estaciones = estaciones[["nombre", "geometry"]]
estaciones = estaciones.to_crs(matchups.crs)
```

```python
estaciones
```

```python
# estacion mas cercana
puntos_cercana = gpd.sjoin_nearest(puntos, estaciones, how="left", distance_col="distancia_m")
puntos_cercana["distancia_km"] = puntos_cercana["distancia_m"] / 1000
```

```python
puntos_cercana
```

```python
puntos_cercana.groupby(["grupo_nombre", "nombre"])["distancia_km"].agg(["count", "min", "mean", "max"]).round(1)
```

```python
centroides = matchups.dissolve(by="grupo_nombre")
centroides = centroides.reset_index()
centroides = centroides[["grupo_nombre", "geometry"]].copy()
centroides["geometry"] = centroides.centroid
```

```python
centroides
```

```python
grupos_cercana = gpd.sjoin_nearest(centroides, estaciones, how="left", distance_col="distancia_m")
grupos_cercana["distancia_km"] = grupos_cercana["distancia_m"] / 1000
grupos_cercana["distancia_km"] = grupos_cercana["distancia_km"].round(1)
grupos_cercana = grupos_cercana.drop(columns=["index_right", "distancia_m"])
```

```python
grupos_cercana
```

```python
estaciones_puntos = pd.DataFrame(estaciones)
estaciones_puntos = estaciones_puntos.rename(columns={"geometry": "geometry_estacion"})
grupos_cercana = grupos_cercana.merge(estaciones_puntos, left_on="nombre", right_on="nombre", how="left")
```

```python
grupos_cercana
```

```python
lineas_geometria = grupos_cercana.shortest_line(gpd.GeoSeries(grupos_cercana["geometry_estacion"], crs=matchups.crs))
lineas = gpd.GeoDataFrame(grupos_cercana[["grupo_nombre", "nombre", "distancia_km"]], geometry=lineas_geometria, crs=matchups.crs)
```

```python
lineas
```

```python
fig, ax = plt.subplots(figsize=(10, 11))

matchups_3857[matchups_3857["grupo_nombre"] == "RN-UPM2"].plot(ax=ax, color="#2a78d6", markersize=10, label="RN-UPM2")
matchups_3857[matchups_3857["grupo_nombre"] == "LDS"].plot(ax=ax, color="#eb6834", markersize=10, label="LDS")
matchups_3857[matchups_3857["grupo_nombre"] == "RN-UPM1"].plot(ax=ax, color="#1baf7a", markersize=10, label="RN-UPM1")
matchups_3857[matchups_3857["grupo_nombre"] == "RDP-MONTES"].plot(ax=ax, color="#eda100", markersize=10, label="RDP-MONTES")

estaciones_3857.plot(ax=ax, color="red", marker="^", markersize=150, edgecolor="black", label="Estaciones INIA")

lineas_3857[lineas_3857["grupo_nombre"] == "RN-UPM2"].plot(ax=ax, color="#2a78d6", linestyle="--", linewidth=2)
lineas_3857[lineas_3857["grupo_nombre"] == "LDS"].plot(ax=ax, color="#eb6834", linestyle="--", linewidth=2)
lineas_3857[lineas_3857["grupo_nombre"] == "RN-UPM1"].plot(ax=ax, color="#1baf7a", linestyle="--", linewidth=2)
lineas_3857[lineas_3857["grupo_nombre"] == "RDP-MONTES"].plot(ax=ax, color="#eda100", linestyle="--", linewidth=2)

ax.annotate(estaciones_3857["nombre"].iloc[0], (estaciones_3857["x"].iloc[0], estaciones_3857["y"].iloc[0]), xytext=(8, 8), textcoords="offset points", fontsize=10, weight="bold")
ax.annotate(estaciones_3857["nombre"].iloc[1], (estaciones_3857["x"].iloc[1], estaciones_3857["y"].iloc[1]), xytext=(8, 8), textcoords="offset points", fontsize=10, weight="bold")
ax.annotate(estaciones_3857["nombre"].iloc[2], (estaciones_3857["x"].iloc[2], estaciones_3857["y"].iloc[2]), xytext=(8, 8), textcoords="offset points", fontsize=10, weight="bold")
ax.annotate(estaciones_3857["nombre"].iloc[3], (estaciones_3857["x"].iloc[3], estaciones_3857["y"].iloc[3]), xytext=(8, 8), textcoords="offset points", fontsize=10, weight="bold")
ax.annotate(estaciones_3857["nombre"].iloc[4], (estaciones_3857["x"].iloc[4], estaciones_3857["y"].iloc[4]), xytext=(8, 8), textcoords="offset points", fontsize=10, weight="bold")

ax.annotate(str(lineas_3857["distancia_km"].iloc[0]) + " km", (lineas_3857["x"].iloc[0], lineas_3857["y"].iloc[0]), xytext=(6, 0), textcoords="offset points", fontsize=10, backgroundcolor="white")
ax.annotate(str(lineas_3857["distancia_km"].iloc[1]) + " km", (lineas_3857["x"].iloc[1], lineas_3857["y"].iloc[1]), xytext=(6, 0), textcoords="offset points", fontsize=10, backgroundcolor="white")
ax.annotate(str(lineas_3857["distancia_km"].iloc[2]) + " km", (lineas_3857["x"].iloc[2], lineas_3857["y"].iloc[2]), xytext=(6, 0), textcoords="offset points", fontsize=10, backgroundcolor="white")
ax.annotate(str(lineas_3857["distancia_km"].iloc[3]) + " km", (lineas_3857["x"].iloc[3], lineas_3857["y"].iloc[3]), xytext=(6, 0), textcoords="offset points", fontsize=10, backgroundcolor="white")

ax.set_xlim(-6550000, -5900000)
ax.set_ylim(-4200000, -3550000)
cx.add_basemap(ax, source=cx.providers.Esri.WorldGrayCanvas)
ax.legend(loc="upper right")
ax.set_axis_off()
plt.show()
```
