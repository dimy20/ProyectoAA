```python
import pandas as pd
import geopandas as gpd
import matplotlib.pyplot as plt
import numpy as np
```

```python
df = pd.read_csv("puntos_examinados.csv")
df_examinados = gpd.GeoDataFrame(
    df,
    geometry=gpd.GeoSeries.from_wkt(df["geometry"]),
    crs="EPSG:4326"
)
df_examinados.head()
```

```python
def check(tipo: str):
    df_tipo = df_examinados[df_examinados["Decision"] == tipo]
    N = df_tipo.shape[0]
    index = np.random.randint(0, N)
    print(df_tipo.iloc[index])
    display(df_tipo.iloc[[index]].explore(
        tiles="Esri.WorldImagery",
        marker_kwds={"radius": 8},
        color="red",
        zoom_start=16,
    ))
```

```python
check("si")
```
