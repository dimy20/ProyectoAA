```python
import pandas as pd
```

```python
data_set = pd.read_csv("../../data/dataset_final/indices_acolite.csv")
print(data_set.shape)
data_set.head()
```

```python
sin_imagen = data_set[data_set["B02_mean"].isna()].copy()
print(sin_imagen.shape)
sin_imagen.head()
```

```python
con_imagen = data_set[data_set["B02_mean"].notna()].copy()
con_imagen.head()
```

```python
con_imagen_40 = con_imagen.sample(40, random_state=42).copy()
con_imagen_40.head()
```

```python
sin_imagen.groupby("estado_trofico").size()
```

```python
sin_imagen_e = sin_imagen[sin_imagen["estado_trofico"] == "E"].copy()
sin_imagen_e.head()
```

```python
sin_imagen_e_20 = sin_imagen_e.sample(20, random_state=42).copy()
sin_imagen_e_20.head()
```

```python
sin_imagen_otros = sin_imagen[sin_imagen["estado_trofico"] != "E"].copy()
sin_imagen_otros.head()
```

```python
sin_imagen_otros_20 = sin_imagen_otros.sample(20, random_state=42).copy()
sin_imagen_otros_20.groupby("estado_trofico").size()
```

```python
sin_imagen_40 = pd.concat([sin_imagen_e_20, sin_imagen_otros_20], ignore_index=True)
sin_imagen_40.groupby("estado_trofico").size()
```

```python
con_datos = con_imagen_40.copy()
con_datos.to_csv("../../data/acolite_prueba/con_datos.csv", index=False)
con_datos.shape
```

```python
sin_datos = sin_imagen_40.copy()
sin_datos.to_csv("../../data/acolite_prueba/sin_datos.csv", index=False)
sin_datos.shape
```
