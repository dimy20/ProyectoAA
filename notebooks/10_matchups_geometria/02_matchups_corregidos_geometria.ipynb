# Matchups finales (con geometria)

Mismo filtro que `06_Datos_Satelitales/matchups_corregidos.ipynb` (RN-UPM1 solo hasta 1 dia de diferencia temporal), aplicado a los matchups de `matchups_geometria.ipynb`.

Salida: `data/nuevos/matchups_registros_finales_geometria.geojson`

```python
import geopandas as gpd
import pandas as pd
```

```python
matchups = gpd.read_file("../../data/nuevos/registros_limpios_matchups_geometria.json")
print(matchups.shape)
matchups.head()
```

```python
matchups["delta"] = pd.to_timedelta(matchups["delta"])
```

```python
matchups_1_dia = matchups[~((matchups["grupo_nombre"] == "RN-UPM1") & (matchups["delta"] > pd.Timedelta(days=1)))].copy()
print(matchups_1_dia.shape)
matchups_1_dia.groupby("grupo_nombre").size()
```

```python
matchups_1_dia["delta"] = (
    matchups_1_dia["delta"].dt.total_seconds()
)
```

```python
matchups_1_dia.to_file("../../data/nuevos/matchups_registros_finales_geometria.geojson",  driver="GeoJSON")
```
