```python
import geopandas as gpd
import pandas as pd
```

```python
matchups = gpd.read_file("../data/registros_limpios_matchups_nuevos.json")
print(matchups.shape)
matchups.head()
```

```python
matchups["delta"] = pd.to_timedelta(matchups["delta"])
```

```python
matchups_3_dias = matchups[~((matchups["grupo_nombre"] == "RN-UPM1") & (matchups["delta"] > pd.Timedelta(days=3)))].copy()
print(matchups.shape)
print(matchups_3_dias.shape)
```

```python
matchups_1_dia = matchups[~((matchups["grupo_nombre"] == "RN-UPM1") & (matchups["delta"] > pd.Timedelta(days=1)))].copy()
print(matchups_1_dia.shape)
```

```python
matchups_2_horas = matchups[~((matchups["grupo_nombre"] == "RN-UPM1") & (matchups["delta"] > pd.Timedelta(hours=2)))].copy()
print(matchups_2_horas.shape)
```

```python
matchups_5_horas = matchups[~((matchups["grupo_nombre"] == "RN-UPM1") & (matchups["delta"] > pd.Timedelta(hours=5)))].copy()
print(matchups_5_horas.shape)
```

```python
clases = pd.DataFrame({
    "1 dia": matchups_1_dia.groupby("estado_trofico").size(),
    "2 horas": matchups_2_horas.groupby("estado_trofico").size(),
    #"5 horas": matchups_5_horas.groupby("estado_trofico").size(),
})
clases["diferencia"] = clases["2 horas"] - clases["1 dia"]
clases.loc["Total"] = clases.sum()
clases
```
