#### Que queremos filtrar ahora?
En el paso anterior lo que hicimos fue filtrar todos los puntos que aparecen en ambas fuentes: OAN y Gems

1. Para cada punto queremos saber que mediciones estan repetidas
2. Una medicion puede estar repitidia si: Fue tomadad en la misma fecha y si el valor es el mismo

```python
mediciones_de_p = mediciones_de_p.set_index("fecha").sort_index()
```

```python
mediciones_de_p.shape
```

```python
N = mediciones_de_p.shape[0]
eps = 1e-5

def fila_extraer_chla(fila, chla_gems_fallback):
    if fila["fuente"] == "GEMS":
        chla = float(fila["value"])
    elif fila["fuente"] == "OAN":
        if fila["valor_transformado"] in casos:
            chla = chla_gems_fallback
        else:
            chla = float(fila["valor_transformado"]) / 1000
    return chla

filas_finales = []
i = 0
ultimo_igual = False
while i < N - 1:
    fecha1 = mediciones_de_p.index[i]
    fecha2 = mediciones_de_p.index[i+1]
    if fecha1 == fecha2:
        print("IGUALES!")
        fila1 = mediciones_de_p.iloc[i]
        fila2 = mediciones_de_p.iloc[i+1]

        if fila1["fuente"] == "GEMS":
            chla_gems = float(fila1["value"])
        else:
            chla_gems = float(fila2["value"])

        chla_f1 = fila_extraer_chla(fila1, chla_gems)
        chla_f2 = fila_extraer_chla(fila2, chla_gems)
        diff = abs(chla_f1 - chla_f2)
        if diff <= eps:
            # GUARDAR UNO SOLO!
            print(chla_f1, chla_f2, "Igual valor")
            filas_finales.append(fila1)
        else:
            print(chla_f1, chla_f2, "Diferente valor")
            filas_finales.append(fila1)
            filas_finales.append(fila2)
        i+=2
        if i == N:
            ultimo_igual = True
    else:
        print("fechas differentes!")
        filas_finales.append(mediciones_de_p.iloc[i])
        i+=1

if not ultimo_igual and N > 0:
    filas_finales.append(mediciones_de_p.iloc[N-1])

mediciones_de_p_sinrep = gpd.GeoDataFrame(pd.DataFrame(filas_finales), geometry="geometry", crs=mediciones_de_p.crs)
```

#### Ahora para todos los puntos
Generalizamos lo anterior: en vez de un solo punto `p`, iteramos sobre todos los puntos de `iguales` y deduplicamos cada uno.

```python
def procesar_punto(p):
    chla_p_gems = gpd.sjoin_nearest(
        p, gems_utm,
        how='inner',
        max_distance=0.1,
        distance_col='dist'
    )

    chla_p_oan = gpd.sjoin_nearest(
        p, oan_utm,
        how='inner',
        max_distance=0.1,
        distance_col='dist'
    )

    mediciones_de_p = gpd.GeoDataFrame(
        pd.concat([chla_p_gems, chla_p_oan], ignore_index=True),
        geometry="geometry",
        crs=oan_utm.crs
    )
    mediciones_de_p = mediciones_de_p.set_index("fecha").sort_index()

    N = mediciones_de_p.shape[0]
    eps = 1e-5

    def fila_extraer_chla(fila, chla_gems_fallback):
        if fila["fuente"] == "GEMS":
            chla = float(fila["value"])
        elif fila["fuente"] == "OAN":
            if fila["valor_transformado"] in casos:
                chla = chla_gems_fallback
            else:
                chla = float(fila["valor_transformado"]) / 1000
        return chla

    filas_finales = []
    i = 0
    ultimo_igual = False
    while i < N - 1:
        fecha1 = mediciones_de_p.index[i]
        fecha2 = mediciones_de_p.index[i+1]
        if fecha1 == fecha2:
            fila1 = mediciones_de_p.iloc[i]
            fila2 = mediciones_de_p.iloc[i+1]

            if fila1["fuente"] == "GEMS":
                chla_gems = float(fila1["value"])
            else:
                chla_gems = float(fila2["value"])

            chla_f1 = fila_extraer_chla(fila1, chla_gems)
            chla_f2 = fila_extraer_chla(fila2, chla_gems)
            diff = abs(chla_f1 - chla_f2)
            if diff <= eps:
                # GUARDAR UNO SOLO!
                filas_finales.append(fila1)
            else:
                filas_finales.append(fila1)
                filas_finales.append(fila2)
            i+=2
            if i == N:
                ultimo_igual = True
        else:
            filas_finales.append(mediciones_de_p.iloc[i])
            i+=1

    if not ultimo_igual and N > 0:
        filas_finales.append(mediciones_de_p.iloc[N-1])

    mediciones_de_p_sinrep = gpd.GeoDataFrame(pd.DataFrame(filas_finales), geometry="geometry", crs=mediciones_de_p.crs)

    return N, mediciones_de_p_sinrep
```

```python
resultados = []
total_in = 0

for idx in range(len(iguales)):
    p = iguales.iloc[[idx]]
    N, mediciones_de_p_sinrep = procesar_punto(p)
    total_in += N
    resultados.append(mediciones_de_p_sinrep)

todas_mediciones_sinrep = gpd.GeoDataFrame(pd.concat(resultados, ignore_index=True), geometry="geometry", crs=oan_utm.crs)
```

```python
todas_mediciones_sinrep
```

```python
total_sin_dedup = total_in
total_con_dedup = todas_mediciones_sinrep.shape[0]

print("Mediciones OAN + GEMS sin deduplicar (puntos compartidos):", total_sin_dedup)
print("Mediciones despues de deduplicar:", total_con_dedup)
print("Duplicados eliminados:", total_sin_dedup - total_con_dedup)
```
