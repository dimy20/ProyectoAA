```python
import os
import sys
from pathlib import Path

import geopandas as gpd
import pandas as pd
from sklearn.model_selection import train_test_split

while not (Path.cwd() / "pyproject.toml").exists():
    os.chdir("..")
sys.path.append(str(Path.cwd()))

from models.core import RANDOM_STATE, TEST_SIZE, GRUPO_SPLIT_DROP_COLUMNS
from models.rfs import SPLITS_DIR
```

```python
FULL_DATASET = "data/chla_indices_v1.geojson"
```

```python
gdf = gpd.read_file(FULL_DATASET)
groups = list(gdf["grupo_nombre"].unique())
```

```python
os.makedirs(SPLITS_DIR, exist_ok=True)
```

```python
for g in groups:
    gdf_grupo = gdf[gdf["grupo_nombre"] == g].copy()
    gdf_grupo = gdf_grupo.drop(columns=GRUPO_SPLIT_DROP_COLUMNS)
    gdf_grupo = pd.DataFrame(gdf_grupo)
    train_df, val_df = train_test_split(gdf_grupo, test_size=TEST_SIZE, stratify=gdf_grupo["estado_trofico"], random_state=RANDOM_STATE)
    grupo_split_dir = SPLITS_DIR / f"{g}"
    os.makedirs(grupo_split_dir, exist_ok=True)
    train_df.to_csv(grupo_split_dir/ "train.csv")
    val_df.to_csv(grupo_split_dir/ "val.csv")
```
